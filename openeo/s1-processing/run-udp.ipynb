This notebook demonstrates how to run the S1 logistic User Defined Process

```python
import logging

import shapely
import openeo

from utils import urls

logging.basicConfig(level=logging.INFO)
```

```python
connection = openeo.connect("openeo.dataspace.copernicus.eu")
```

```python
connection.authenticate_oidc()
```

```python
# very small test AOI
x = 30.944
y = 1.273
delta = 0.1
spatial_extent = {
    "west": x,
    "south": y,
    "east": x + delta,
    "north": y + delta,
}
```

```python
# spatial extent as dict of Polygon geometry
spatial_extent = shapely.geometry.mapping(
    shapely.box(
        xmin=spatial_extent["west"],
        ymin=spatial_extent["south"],
        xmax=spatial_extent["east"],
        ymax=spatial_extent["north"],
    )
)
```

```python
cube = connection.datacube_from_process(
    "s1_logistic_processing",
    namespace=urls.S1_PROCESSING_UDP,
    # spatial_extent=spatial_extent,
)
```

```python
job = cube.create_job(out_format="GTiff")
# job = cube.create_job(out_format="netCDF")
```

```python
job.start_and_wait()

# full run cost 135 credits
```

```python
results = job.get_results()
```
